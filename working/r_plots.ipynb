```r
# Read the CSV file 'google_serp_top30_from_json.csv' located in working/data/
google_data <- read.csv("/Users/benkan45d6/GitHub/Intro-To-DataScience/working/data/google_serp_top30_from_json.csv", stringsAsFactors = FALSE)
head(google_data)
```

```r
# Plot side-by-side bar chart of restaurant ratings from Google and OpenRice (on same 0–5 scale), sorted by OpenRice rating descending, y-axis limited to 3–5, and plot is wide for readability
library(ggplot2)
library(tidyr)
library(dplyr)

# The available relevant columns for ratings:
# "restaurant", "google_rating", "openrice_smile", "openrice_cry"

# Ensure openrice_smile and openrice_cry are numeric
google_data$openrice_smile <- as.numeric(google_data$openrice_smile)
google_data$openrice_cry <- as.numeric(google_data$openrice_cry)

# Calculate OpenRice rating (smile / (smile + cry) * 5)
google_data$openrice_rating_converted <- with(
  google_data, 
  ifelse(
    is.na(openrice_smile) | is.na(openrice_cry) | (openrice_smile + openrice_cry) == 0, 
    NA, 
    openrice_smile / (openrice_smile + openrice_cry) * 5
  )
)

# Prepare and sort data: select columns, sort by OpenRice rating desc (no row cap)
plot_data <- google_data %>%
  select(restaurant, google_rating, openrice_rating_converted) %>%
  arrange(desc(openrice_rating_converted)) %>%
  rename(OpenRice = openrice_rating_converted, Google = google_rating)

plot_data$restaurant <- factor(plot_data$restaurant, levels = plot_data$restaurant)

# Pivot longer for plotting
plot_data_long <- plot_data %>%
  pivot_longer(
    cols = c(Google, OpenRice),
    names_to = "Platform",
    values_to = "Rating"
  )

# Create a wider plot by setting figure size before plotting (works in most notebooks/RStudio/IRkernel etc.)
if (exists("IRdisplay::display_html")) { # for IRkernel
  IRdisplay::display_html("<style>div.output_scroll {height:auto; max-height:1000px;}</style>")
}
options(repr.plot.width=16, repr.plot.height=6) # for Jupyter/IRkernel, does nothing in base R

# Plot side-by-side bars (both on 0–5 scale, ordered by OpenRice rating descending), y-axis limited to 3–5
ggplot(plot_data_long, aes(x = restaurant, y = Rating, fill = Platform)) +
  geom_bar(stat = "identity", position = position_dodge(width = 0.8)) +
  labs(
    title = NULL,
    x = "Restaurant",
    y = "Rating (0–5 scale)"
  ) +
  theme(
    axis.text.x = element_text(angle = 45, hjust = 1, size = 10),
    plot.margin = margin(10, 30, 10, 10),
    plot.title = element_text(size = 16),
    legend.position = "top"
  ) +
  scale_fill_manual(values = c("steelblue", "orange")) +
  coord_cartesian(ylim = c(3, 5))
```

```r
# Paired t-test: is OpenRice systematically higher than Google?
# Only consider rows with both ratings present (no NAs)
ttest_data <- plot_data %>% filter(!is.na(OpenRice) & !is.na(Google))
ttest_result <- t.test(ttest_data$OpenRice, ttest_data$Google, paired = TRUE)

cat("Paired t-test: OpenRice vs Google ratings\n")
cat("-----------------------------------------\n")
cat(sprintf("t = %.3f, df = %d, p-value = %.4g\n", ttest_result$statistic, ttest_result$parameter, ttest_result$p.value))
cat(sprintf("Mean of differences: %.3f\n", mean(ttest_data$OpenRice - ttest_data$Google, na.rm = TRUE)))
cat(sprintf("95%% CI: [%.3f, %.3f]\n", ttest_result$conf.int[1], ttest_result$conf.int[2]))
cat(sprintf("Alternative hypothesis: %s\n", ttest_result$alternative))
```
